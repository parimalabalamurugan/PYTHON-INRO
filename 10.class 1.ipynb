```python
class MultipleFunction:

  
    def add(a, b):
        return a + b


    def odd_even():
        num = int(input("Enter a number: "))
        if num % 2 == 0:
            print(num, "is Even Number")
        else:
            print(num, "is Odd Number")


# Calling add function
result = MultipleFunction.add(5, 1)
print("add =", result)

# Calling odd_even function
MultipleFunction.odd_even()
```
