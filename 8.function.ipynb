```python
#function  creation and calling

def AgeCategory():
    lists = [20, 30, 40, 50]

    for age in lists:
        if age < 18:
            print("child")
        elif age < 35:
            print("adult")
        elif age < 59:
            print("citizen")
        else:
            print("senior")

AgeCategory()
```
